Extra 1: Control de aforo con break y continue (while)
En este ejercicio utilicé un ciclo while para registrar el costo de los boletos de varios visitantes.
Entrada: Cantidad de visitantes y costo de cada boleto.
Proceso: Se van sumando los costos y si el boleto es de $0 se muestra que el menor no tiene cargo y se continúa con el siguiente visitante.
Salida:Se muestra el total acumulado de los boletos.

```python
#JULIO CESAR ANGELES MENDOZA (IDS)
cantidad_visitantes = int(input("Coloca la cantidad de visitantes: "))
contador = 1
total = 0
 
while contador <= cantidad_visitantes:
    costo = float(input(f"costo del boleto:{contador}")) 
    if costo == 0:
        print("Menor de 3 años, sin cargo")
    total += costo
    contador = contador + 1
    if total >= 100:
        print("Aforo alcanzado se detiene el registro")
        print(f"Total acomulado: ${total:.2f}")
        continue
        break
    
    

 
```

Extra 2: Estadística de visitantes con for
En este ejercicio utilicé un ciclo for para registrar las edades de los visitantes y obtener algunos datos.
Entrada: Cantidad de visitantes y edad de cada uno.
Proceso:Se suman las edades, se cuentan las personas mayores de edad y se calcula el promedio.
Salida:Se muestra la cantidad de adultos y el promedio de edad de los visitantes.

```python
#JULIO CESAR ANGELES MENDOZA (IDS)
cantidad_visitantes = int(input("Coloca la cantidad de visitantes: "))
adultos = 0
suma_edades = 0

for contador in range (cantidad_visitantes):
    edad = int(input("ingresa tu edad: "))
    suma_edades += edad
    if edad >= 18:
        adultos += 1

promedio_edad = suma_edades / cantidad_visitantes

print(f"adultos {adultos}")
print(f"Promedio de edad: {promedio_edad:.2f}")
 
```

En este ejercicio utilicé ciclos for anidados para crear una pirámide de asteriscos.
Entrada: Altura de la pirámide.
Proceso: Un ciclo controla las filas y otro la cantidad de asteriscos que se imprimen.
Salida: Se muestra la pirámide con la altura indicada.

```python
#JULIO CESAR ANGELES MENDOZA (IDS)
altura = int(input("Altura de la pirámide: "))

for fila in range(1, altura + 1):
    for asterisco in range(fila):
        print("*", end="")
    print()
```
