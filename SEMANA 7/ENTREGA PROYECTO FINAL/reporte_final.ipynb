1. . Introducción

El presente proyecto consiste en el desarrollo de un sistema en Python denominado Control de Ocupación del Gimnasio Universitario. Su propósito es llevar un registro básico de los usuarios que utilizan las instalaciones, controlar sus entradas y salidas y determinar la cantidad de personas que permanecen dentro del gimnasio en una fecha determinada.

Para su desarrollo se utilizaron los conocimientos adquiridos durante la materia de Fundamentos de Programación, entre ellos variables, cadenas de texto, listas, matrices, tuplas, estructuras condicionales, ciclos for y while, funciones, archivos de texto, manejo de excepciones y depuración mediante el módulo PDB.

El sistema utiliza archivos .txt para conservar la información, por lo que los datos registrados permanecen disponibles aun después de cerrar el programa.

2. Problemática de negocio

Un gimnasio universitario recibe diariamente diferentes usuarios. Si no existe un sistema de control pueden presentarse problemas como desconocer cuántas personas se encuentran dentro de las instalaciones, registrar entradas o salidas de manera incorrecta, permitir accesos de personas no registradas, duplicar entradas, registrar salidas sin una entrada previa o perder la información al finalizar el programa.

Por este motivo se desarrolló un sistema que permite registrar usuarios, controlar sus entradas y salidas y conservar la información mediante archivos de texto.

3. Objetivo general

Desarrollar un sistema en Python que permita registrar y controlar la ocupación de un gimnasio universitario mediante la administración de usuarios, entradas y salidas, incorporando validaciones, persistencia de datos, manejo de excepciones y mecanismos de depuración.

4. Objetivos específicos

El sistema permite identificar al operador mediante un nombre o nickname, mostrar una bienvenida personalizada, utilizar una pantalla de carga, presentar las funciones mediante un menú, detectar periodos de inactividad, validar fechas, almacenar datos en archivos de texto, manejar errores de ejecución y utilizar PDB para revisar la lógica del programa.

5. Alcance del sistema

El proyecto funciona como un prototipo académico de control de ocupación. Permite registrar usuarios, entradas y salidas, consultar la ocupación, leer archivos, crear nuevos archivos y anexar información.

No se incluyen funciones como pagos, membresías, horarios, sensores, torniquetes, bases de datos o control biométrico, debido a que estos elementos se encuentran fuera del alcance del proyecto.

7. Reglas de negocio

El sistema utiliza diferentes reglas para mantener la congruencia de la información.

El operador debe ingresar un nombre o nickname antes de acceder al sistema.

Los campos importantes no pueden permanecer vacíos.

Un usuario debe encontrarse registrado antes de poder ingresar al gimnasio.

No se permite registrar dos veces al mismo usuario.

No se permite registrar una segunda entrada si el usuario ya se encuentra dentro.

No se permite registrar una salida si el usuario no tiene una entrada activa.

La fecha debe existir realmente y utilizar el formato dd/mm/aaaa.

No se permiten fechas anteriores al inicio de operaciones ni fechas futuras.

Los registros deben permanecer almacenados después de cerrar el programa.

Los errores relacionados con archivos deben ser controlados sin provocar el cierre inesperado del sistema.

La ocupación se obtiene restando las salidas registradas a las entradas correspondientes a la fecha seleccionada.

8. Identificación del usuario

Al iniciar el sistema se solicita al operador un nombre o nickname.

Ejemplo:

Ingresa tu nombre o nickname: Raul

Si el usuario deja el campo vacío, el programa muestra un mensaje de error y vuelve a solicitar la información.

9. Bienvenida dinámica

Después de ingresar el nombre se muestra una bienvenida personalizada.

Ejemplo:

¡Hola Raul! ¡BIENVENIDO AL SISTEMA DE CONTROL DE OCUPACIÓN DEL GIMNASIO UNIVERSITARIO!

Esto se realiza mediante cadenas de texto y f-strings.

10. Pantalla de carga

Antes de mostrar el menú principal se ejecuta una pantalla de carga.

La función utiliza time.sleep() y un ciclo for para mostrar puntos de forma progresiva.

Ejemplo:

Cargando sistema.....
Sistema cargado correctamente.

11. Menú como matriz

Las opciones del menú se almacenan mediante una lista de listas, lo cual permite representarlo como una matriz.

Las opciones disponibles son:

1. Registrar usuario
2. Registrar entrada
3. Registrar salida
4. Consultar ocupación
5. Leer archivo
6. Crear archivo
7. Escribir o anexar datos
8. Salir

El programa utiliza un ciclo for para recorrer e imprimir cada fila del menú.

12. Control de inactividad

El sistema cuenta con un control de inactividad de 10 minutos.

La constante utilizada es:

TIEMPO_INACTIVIDAD = 600

El programa utiliza un ciclo for para contar los segundos sin interacción.

Si el usuario presiona una tecla, el tiempo vuelve a comenzar.

Si transcurren 10 minutos sin actividad, aparece el siguiente mensaje:

Han transcurrido 10 minutos sin actividad.
¿Deseas continuar en el menú? Escribe si o no:

Si la respuesta es si, el usuario permanece en el menú.

Si la respuesta es no, el programa regresa a la pantalla inicial.

13. Captura y validación de fecha

El programa solicita una fecha de operación mediante el formato:

dd/mm/aaaa

La fecha es validada mediante datetime.strptime(), lo cual permite comprobar que exista realmente.

Por ejemplo, una fecha como 31/02/2026 no es aceptada.

Posteriormente la fecha se almacena como una tupla:

Fecha = (dia, mes, anio)

Ejemplo:

Fecha = (25, 9, 2026)

También se valida que no sea anterior al inicio de operaciones ni posterior a la fecha actual.

14. Registro de usuarios

La opción Registrar usuario solicita el nombre de una persona y verifica que el campo no esté vacío y que el usuario no se encuentre registrado anteriormente.

Cuando la información es válida se almacena en usuarios.txt.

Ejemplo:

Fecha: 25/09/2026 | Usuario: Ana Maria

El programa también realiza una comparación del nombre completo para evitar confundir nombres similares.

Por ejemplo:

Ana
Ana Maria
Ana Paula

son tratados como usuarios diferentes.

15. Registro de entradas

Antes de permitir una entrada, el sistema verifica que el usuario se encuentre registrado.

También comprueba que no tenga otra entrada activa.

Cuando las validaciones son correctas se agrega el registro a entradas.txt.

Ejemplo:

Fecha: 25/09/2026 | Entrada: Ana Maria

16. Registro de salidas

Para registrar una salida, el usuario debe encontrarse registrado y tener una entrada activa.

Esto evita registrar salidas de personas que no se encuentran dentro del gimnasio.

Ejemplo:

Fecha: 25/09/2026 | Salida: Ana Maria

17. Cálculo de ocupación

La ocupación se obtiene mediante la siguiente relación:

Ocupación = Entradas - Salidas

El programa revisa los registros correspondientes a la fecha seleccionada y cuenta las entradas y salidas.

Ejemplo:

Entradas: 8
Salidas: 5
Ocupación actual: 3 personas

Si el resultado fuera negativo, el sistema muestra cero para evitar una ocupación inválida.

Cada consulta también se almacena en ocupacion.txt para conservar un historial.

18. Persistencia de datos

Una parte importante del proyecto es la persistencia.

Los archivos utilizados son:

usuarios.txt
entradas.txt
salidas.txt
ocupacion.txt

El programa utiliza diferentes modos de apertura.

"r" se utiliza para lectura.

"a" se utiliza para anexar información sin borrar lo anterior.

"x" se utiliza para crear archivos nuevos evitando sobrescribir archivos existentes.

Gracias a esto, los datos permanecen disponibles aunque el programa sea cerrado.

19. Lectura de archivos

El sistema permite seleccionar y consultar los archivos disponibles desde el menú.

Si el nombre del archivo es incorrecto, el sistema muestra un mensaje de error y continúa ejecutándose.

Cuando el archivo existe, su contenido se muestra directamente en consola.

20. Creación de archivos

El programa permite crear nuevos archivos .txt.

Antes de hacerlo valida que el nombre no esté vacío y que no contenga caracteres inválidos.

También controla el caso en que el archivo ya existe.

Los archivos creados son incorporados al diccionario de archivos disponibles.

21. Escritura y anexado

El sistema permite seleccionar un archivo existente y agregar nueva información.

La información se almacena junto con la fecha seleccionada.

El contenido anterior no se elimina porque se utiliza el modo de apertura "a".

22. Control de excepciones

Se implementaron bloques try-except para evitar que errores relacionados con los archivos finalicen inesperadamente el programa.

Las principales excepciones utilizadas son:

FileNotFoundError: controla archivos inexistentes.

FileExistsError: controla intentos de crear un archivo que ya existe.

PermissionError: controla problemas de permisos.

OSError: controla otros errores relacionados con archivos o con el sistema operativo.

Además, el programa valida fechas, nombres, respuestas y opciones del menú.

Gracias a estas validaciones, si el usuario ingresa datos incorrectos el sistema muestra un mensaje de error y continúa ejecutándose.

23. Depuración con PDB

Para realizar la depuración técnica se utilizó el módulo estándar pdb.

El programa contiene la variable:

DEBUG = False

Para realizar pruebas puede cambiarse a:

DEBUG = True

Cuando está activo se ejecutan puntos de interrupción mediante:

pdb.set_trace()

Durante la depuración se pueden revisar variables como:

Fecha
opcion
nombre
dentro

Algunos comandos utilizados son:

p variable: muestra el valor de una variable.
n: ejecuta la siguiente línea.
c: continúa con la ejecución.
q: termina la sesión de depuración.

24. Fallas localizadas durante el desarrollo

Durante el desarrollo se detectaron y corrigieron diferentes problemas.

Uno de los primeros problemas fue que algunas funciones necesitaban recibir la fecha pero inicialmente eran llamadas sin ese argumento.

También se corrigió la validación de fechas para evitar valores inexistentes.

Se mejoró el flujo del menú para regresar correctamente a la pantalla inicial después del periodo de inactividad.

Se agregaron controles para archivos inexistentes, errores de permisos y archivos duplicados.

También se corrigió la comparación de nombres. Inicialmente una búsqueda parcial podía provocar que Ana y Ana Maria fueran considerados el mismo usuario. Posteriormente se modificó la lógica para comparar el nombre completo.

Finalmente se agregaron validaciones para impedir entradas duplicadas y salidas sin una entrada previa.

29. Conclusión

El desarrollo del Sistema de Control de Ocupación del Gimnasio Universitario permitió aplicar de manera práctica los principales conocimientos estudiados durante la materia de Fundamentos de Programación.

El sistema permite registrar usuarios, controlar entradas y salidas, calcular la ocupación de una fecha y conservar la información mediante archivos de texto.

Las validaciones implementadas ayudan a mantener congruencia entre los registros y evitan operaciones como usuarios duplicados, entradas repetidas o salidas sin una entrada previa.

El manejo de excepciones permite que los errores relacionados con archivos no interrumpan completamente la ejecución.

Finalmente, la utilización de PDB permitió revisar variables y localizar fallas de lógica durante el desarrollo.

El resultado es un prototipo funcional que cumple con los requerimientos establecidos y demuestra la aplicación de estructuras de control, funciones, matrices, tuplas, archivos, excepciones, persistencia y herramientas de depuración en Python.
