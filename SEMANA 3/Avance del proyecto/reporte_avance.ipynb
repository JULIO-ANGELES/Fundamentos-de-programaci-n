1. ------------ANALISIS ORGANIZACIONAL:-------------
El presente proyecto se desarrolla dentro del contexto de la institución universitaria, específicamente en el área que brinda el servicio de gimnasio para la comunidad TecMilenio ,Campus Queretaro.

El gimnasio representa un espacio de uso común para estudiantes y personal de la universidad. Su nivel de ocupación puede variar considerablemente dependiendo del horario, generando periodos donee hay pocos usuarios asi como tambien donde se satura bastante de estos mismos.

Esta variación puede dificultar que los usuarios conozcan cuáles son los horarios con mayor o menor demanda de usuarios, de igual manera, el personal responsable del gimnasio puede requerir información sencilla que le permita saber las tendencias de ocupación y reconocer los periodos en los que existe una mayor concentración de usuarios.

Analizando esta problematica desde la perspectiva de un usuario del gimnasio, propongo un programa orientado al registro y análisis de la ocupación del gimnasio universitario.

El programa permitirá ingresar la cantidad de usuarios presentes en diferentes horarios, calcular el porcentaje de ocupación con respecto a la capacidad máxima, determinar los espacios disponibles y clasificar el nivel de ocupación de acuerdo con parámetros previamente establecidos.

La información generada permitirá identificar los horarios de mayor y menor demanda, proporcionando una referencia que pueda ser utilizada tanto para el análisis operativo del gimnasio como para conocer los periodos en los que existe menor saturación. asi a que tanto como personal y usuarios usen los datos de forma que mejor les convenga.

AREA DE IMPACTO:
El area de impacto selecionada abarca el gimnasio de la universidad que entra dentro de los servicios deportivos de esta misma.

NECESIDADES QUE SE CONSIDERAN:

Conocer la ocupación del gimnasio.
Detectar horarios de alta demanda.
Identificar áreas saturadas.
Saber cuántos espacios permanecen disponibles.
Generar información sencilla para usuarios y responsables del gimnasio.

2. --------------DEFINICION DEL PROBLEMA Y REGLAS DEL NEGOCIO:----------
2.1 DEFINICION DEL PROBLEMA
La falta de un mecanismo sencillo para registrar y analizar estos niveles de ocupación dificulta identificar rápidamente cuáles son los horarios de mayor y menor demanda. haciendo que los usuarios acudan en periodos de alta afluencia sin contar previamente con una referencia sobre el nivel de ocupación, mientras que la administración dispone de información limitada para analizar el comportamiento de la demanda durante el día.

A partir de los datos registrados, el programa calculará el porcentaje de ocupación y los espacios disponibles, clasificará el nivel de afluencia y permitirá identificar los horarios con mayor y menor ocupación.

En esta primera fase, los datos serán capturados manualmente por el usuario. El proyecto se enfocará en desarrollar y comprobar la lógica necesaria para analizar la ocupación, sin contemplar todavía la conexión con torniquetes, sensores, bases de datos u otros programas externos.

2.2 REGLAS DEL NEGOCIO:
Para el funcionamiento del codigo se tendran en cuenta las siguientes reglas:

R1. Capacidad máxima: El gimnasio deberá contar con una capacidad máxima de usuarios definida antes de comenzar el registro de ocupación.

R2. Registro por horario: Cada registro deberá estar asociado con un horario y con la cantidad de usuarios presentes durante ese periodo.

R3. Validación de usuarios: La cantidad de usuarios registrada no podrá ser menor a cero ni superior a la capacidad máxima del gimnasio.

R4. Espacios disponibles: Los espacios disponibles se determinarán restando los usuarios presentes a la capacidad máxima.

R5. Porcentaje de ocupación: El nivel porcentual de ocupación se calculará mediante la relación entre los usuarios presentes y la capacidad máxima del gimnasio.

R6. Clasificación de ocupación: Para efectos del prototipo, los niveles de ocupación se clasificarán de acuerdo con los siguientes parámetros internos:
De 0% a 50%: "ocupación baja".
Mayor a 50% y hasta 75%: "ocupación moderada".
Mayor a 75% y hasta 90%: "ocupación alta".
Mayor a 90% y hasta 100%: "ocupación crítica".

R7. Capacidad completa: Cuando la cantidad de usuarios sea igual a la capacidad máxima, el programa deberá indicar que no existen espacios disponibles.

R8. Múltiples registros: El programa permitirá capturar diferentes horarios durante una misma ejecución para posteriormente analizar la información registrada.

R9. Análisis de demanda: El programa deberá identificar el horario que presente la mayor ocupación y el horario que presente la menor ocupación entre los registros capturados.

R10. Ocupación promedio: El programa calculará el promedio de ocupación de los horarios registrados para proporcionar un indicador general del uso de las instalaciones.

Estas reglas delimitan el funcionamiento del codigo de este proyecto y servirán como base para hacer despues el algoritmo, pseudocódigo y prototipo en python.

3. ----------Listado de Requerimientos Funcionales--------------

A partir de la problemática y las reglas de negocio establecidas, se definen los siguientes requerimientos funcionales para el programa que hara mas sencillo lidiar con esta problematica

1. Registrar capacidad máxima: El programa deberá permitir ingresar la capacidad máxima de usuarios permitida en el gimnasio.

2. Mostrar menú principal: El programa deberá mostrar un menú que permita al usuario seleccionar las diferentes operaciones disponibles.

3. Registrar ocupación por horario: El programa deberá permitir ingresar un horario y la cantidad de usuarios presentes durante ese periodo.

4. Validar cantidad de usuarios: El programa deberá comprobar que la cantidad registrada sea igual o mayor a cero y que no supere la capacidad máxima del gimnasio.

5. Calcular espacios disponibles: El programa deberá calcular la cantidad de espacios disponibles mediante la diferencia entre la capacidad máxima y los usuarios registrados.

6. Calcular porcentaje de ocupación: El programa deberá determinar el porcentaje de ocupación correspondiente a cada horario registrado.

7. Clasificar el nivel de ocupación: El programa deberá clasificar cada registro como ocupación baja, moderada, alta o crítica, utilizando los parámetros establecidos en las reglas de negocio.

8. Generar alertas de capacidad: Cuando el gimnasio alcance su capacidad máxima, el programa deberá indicar que no existen espacios disponibles.

9. Registrar múltiples horarios: El programa deberá permitir realizar varios registros de ocupación durante una misma ejecución.

10. Consultar registros: El usuario deberá poder visualizar los horarios registrados junto con su cantidad de usuarios y porcentaje de ocupación.

11. Calcular ocupación promedio: El programa deberá calcular el promedio de ocupación utilizando los registros disponibles.

12. Identificar mayor y menor ocupación: El programa deberá identificar los horarios que presenten el mayor y menor nivel de ocupación entre los registros realizados.

13. Mostrar resumen: El programa deberá presentar un resumen con los principales resultados obtenidos durante el análisis.

14. Finalizar el programa: El usuario deberá disponer de una opción para terminar correctamente la ejecución del programa.


4. -----------CLASIFICACION DE DATOS-----------------

Para el desarrollo del programa será necesario utilizar diferentes tipos de datos que permitan almacenar y procesar correctamente la información relacionada con la ocupación del gimnasio.

Los principales tipos de datos utilizados serán enteros (int), números decimales (float), cadenas de texto (str) y valores booleanos (bool).
El tipo int se utilizará para cantidades que se representan mediante números enteros, como la capacidad máxima, el número de usuarios, los espacios disponibles y las opciones del menú.
El tipo float será utilizado para los resultados que pueden contener decimales, principalmente los porcentajes y promedios de ocupación.
El tipo str permitirá almacenar información textual, como el horario registrado y la clasificación del nivel de ocupación.


5. ------------ Operadores del lenguaje-----------

Este proyecto va a utilizar operadores matemáticos, relacionales y lógicos para realizar los cálculos y tomar decisiones a partir de los datos registrados en la entrada.

Operadores matemáticos: Se utilizarán +, -, / y *.
Por ejemplo:
espacios_disponibles = capacidad_maxima - usuarios_presentes
porcentaje_ocupacion = (usuarios_presentes / capacidad_maxima) * 100

La suma también permitirá acumular los porcentajes registrados para posteriormente obtener el promedio de ocupación.

Operadores relacionales: Se utilizarán >, <, >=, <=, == y != para comparar los datos.
Por ejemplo:
usuarios_presentes <= capacidad_maxima

permitirá comprobar que el número de usuarios no supere la capacidad del gimnasio.
También se utilizarán para comparar el porcentaje obtenido con los rangos establecidos para cada nivel de ocupación.

Operadores lógicos: Se utilizarán principalmente and, or y not para combinar condiciones.
Por ejemplo:
usuarios_presentes >= 0 and usuarios_presentes <= capacidad_maxima

permitirá validar que el número de usuarios se encuentre dentro del rango permitido.
