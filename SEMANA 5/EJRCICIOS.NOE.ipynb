```python
candidatos ={
    "A": 0,
    "B": 0,
    "C": 0
}

cantidad = int(input("Cuantos van a ser?"))
conteo = 1

while conteo <= cantidad: 
    voto = input(f"ingrese el voto {conteo} (A, B o C): ")
    print(f"votaste por {voto.upper()} ")
    if voto.upper() in candidatos:
        print(f"el voto para {voto.upper()}")
        candidatos[voto.upper()] += 1
        conteo += 1
    else:
        print("el voto no es valido")

print("\tRESULTADOS:")
for k,v in candidatos.items():
    print(f"opcion {k}, votos: {v}")

```
