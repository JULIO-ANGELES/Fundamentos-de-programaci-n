En este reto se integran tuplas, diccionarios, excepciones y strings dentro de un menú principal. Cada tema se desarrolla en una función diferente para mantener el programa ordenado y permitir que el usuario seleccione qué sección ejecutar.

```python

def seccion_tuplas():
    numeros = (8, 3, 10, 5, 7)

    print("Tercer elemento:", numeros[2])

    numero1 = float(input("Ingresa un numero adicional: "))
    numero2 = float(input("Ingresa otro numero adicional:"))

    nueva_tupla = numeros + (numero1, numero2)

    lista_numeros = list(nueva_tupla)
    lista_numeros.sort()


    print("Nueva tupla: ", nueva_tupla)
    print("Lista ordenada:", lista_numeros)

    suma_total = sumar_tupla(nueva_tupla)
    print("Suma total:", suma_total)

def sumar_tupla(numeros):
    suma = 0

    for numero in numeros:
        suma = suma + numero

    return suma


def seccion_diccionarios():
    contactos = {
        "Ana": "555-0101",
        "Luis": "555-0102",
        "Mia": "555-0103"
    }

    nombre_nuevo = input("Nombre del nuevo contacto: ")
    telefono_nuevo = input("Telefono del nuevo contacto: ")

    contactos[nombre_nuevo] = telefono_nuevo

    print("Contactos registrados:")

    for nombre in contactos.keys():
        print(nombre)

    nombre_buscar = input("Nombre a buscar: ")

    telefono = buscar_contacto(contactos, nombre_buscar)

    if telefono != None:
        print("Telefono:", telefono)
    else:
        print("Contacto no encontrado")


def buscar_contacto(contactos, nombre):
    if nombre in contactos:
        return contactos[nombre]
    else:
        return None


def seccion_excepciones():
    try:
        num1 = int(input("Ingresa el primer numero entero: "))
        num2 = int(input("Ingresa el segundo numero entero: "))

        suma = num1 + num2
        print("Suma:", suma)

        division = num1 / num2
        print("Division:", division)

    except ValueError:
        print("ERROR: debes ingresar numeros enteros")

    except ZeroDivisionError:
        print("ERROR: no se puede dividir entre cero")


def seccion_strings():
    mensaje = "Python es un lenguaje poderoso"

    print("Longitud:", len(mensaje))

    mayusculas = mensaje.upper()
    print("Mayusculas:", mayusculas)

    reemplazado = mensaje.replace("Python", "programacion")
    print("Texto reemplazado:", reemplazado)

    total_palabras = contar_palabras(mensaje)
    print("Palabras totales:", total_palabras)

def contar_palabras(mensaje):
    palabras = mensaje.split()
    return len(palabras)


while True:

    print("\n===== MENU PRINCIPAL =====")
    print("1. Tuplas")
    print("2. Diccionarios")
    print("3. Excepciones")
    print("4. Strings")
    print("5. Finalizar")

    try:
        opcion = int(input("Selecciona una opcion: "))

        match opcion:

            case 1:
                seccion_tuplas()

            case 2:
                seccion_diccionarios()

            case 3:
                seccion_excepciones()

            case 4:
                seccion_strings()

            case 5:
                print("Programa finalizado")
                break

            case _:
                print("Opcion no valida")

    except ValueError:
        print("Debes ingresar una opcion numerica")
```
