EXTRA 1: Sistema de calificaciones con tuplas
Se usa una tupla para guardar las calificaciones, se convierte en lista para ordenarla y una función calcula la suma total.

```python
def sumar_personas(lista):
    suma = 0

    for calificacion in lista:
        suma = suma + calificacion

    return suma




calificaciones = (7.5, 9.0, 8.0, 6.5, 10.0)

nueva_calificacion1 = float(input("Nueva calificacion 1: "))
nueva_calificacion2 = float(input("Nueva calificacion 2: "))

nuevas_calificaciones = (nueva_calificacion1, nueva_calificacion2)

nueva_tupla = calificaciones + nuevas_calificaciones

print("Tercera Calificacion:", calificaciones[2])
print("nueva tupla:", nueva_tupla)

lista_de_nombres = list(nueva_tupla)
lista_de_nombres.sort(reverse=True)

print("Lista ordenada(mayor a menor):", lista_de_nombres)
suma_total = sumar_personas(lista_de_nombres)

print("Suma total:", suma_total)

```

Extra 2: Agenda de contactos con búsqueda
Se usa un diccionario para relacionar cada nombre con su teléfono. Se agrega un contacto y una función permite buscar su número.

```python

agenda = {
    "Ana": "555-0101",
    "Luis": "555-0102",
    "Mía": "555-0103"
}


Nombre = input("Nombre del nuevo contacto:")
Numero = input("Ingresa el Número del nuevo contacto:")

agenda[Nombre] = Numero

# Busca el numero de telefono
def obtener_telefono(nombre, diccionario_contactos):
    return diccionario_contactos.get(nombre, "Contacto no encontrado")

# Aqui se imprime el Nuevo Diccionario solo con nombres gracias a ".keys".
print(f"Contactos registrados: {', '.join(agenda.keys())}")

telefono_encontrado = obtener_telefono(Nombre, agenda)
print(f"El teléfono de {Nombre} es: {telefono_encontrado}")
```

Desafío Extra 2: Contador de votos con diccionario
Crea un diccionario `votos = {"Rojo": 0, "Azul": 0, "Verde": 0}`. Pide al usuario
el color de su voto cinco veces, suma uno al valor correspondiente en cada ocasión
y, al final, imprime con `items()` el color ganador junto con su total

Se usa un diccionario para guardar cada color y sus votos. Al final se muestran los resultados y el ganado

```python
votos ={
    "ROJO": 0,
    "AZUL": 0,
    "VERDE": 0
}

cantidad = int(input("Cuantos van a ser?"))
conteo = 1

if cantidad > 5:
    print("solo se pueden agregar maximo 5 votos")
else:
    while conteo <= cantidad: 
        voto = input(f"ingrese el voto {conteo} (ROJO, AZUL o VERDE): ")
        print(f"votaste por {voto.upper()} ")
        if voto.upper() in votos:
            print(f"el voto para {voto.upper()}")
            votos[voto.upper()] += 1
            conteo += 1
        else:
            print("el voto no es valido")

    print("\tRESULTADOS:")
    for k,v in votos.items():
        print(f"opcion {k}, votos: {v}")
        ganador = max(votos, key=votos.get)

    print("El ganador es:", ganador)
```

Desafío Extra 3: Validación robusta de entrada
Escribe un programa que pida al usuario dos números enteros dentro de un bloque
`try-except`. Si el usuario ingresa texto o un valor vacío, muestra un mensaje
controlado y repite la solicitud con un ciclo `while`. Luego pide la división del
primero entre el segundo, con un bloque específico para `ZeroDivisionError`

Se usa while y try-except para repetir la captura hasta recibir números válidos y controlar la división entre cero.

```python
while True:
    try: 
        num1 = int(input("ingresa el primer numero: "))
        num2 = int(input("ingresa el segundo numero: "))
        break

    except ValueError:
        print("ERROR: debes ingresar un numero entero")
        print("Intenta de nuevo")

try:
    division = num1 / num2
    print("El resultado de la division es: ", division)

except ZeroDivisionError:
    print("ERROR: el numero no es divisible entre cero")
    print("ingresa un nuevo divisor")
```

Desafío Extra 5: Minimenú modular integrador
Construye un menú con un ciclo `while` y opciones numéricas que permita ejecutar de
forma modular: (1) mostrar la tupla de números ordenada, (2) buscar un teléfono en
el diccionario `contactos`, (3) dividir dos números con manejo de
`ZeroDivisionError` y (4) analizar un mensaje de texto. La opción (5) debe
finalizar el programa.

Se utiliza while para repetir el menú y funciones para separar cada opción. El programa termina cuando se selecciona salir.

```python
def ordenar_numeros():
    numeros = (8,3,10,2,7,5)

    lista_numeros = list(numeros)
    lista_numeros.sort()

    print("Numeros ordenados: ", lista_numeros)

def buscar_telefono():
    contactos = {
        "RAUL": "4421234567",
        "CARLOS": "4429876543",
        "ANA": "4424567890"
    }

    nombre = input("Ingrese el nombre de contactos:  ")
    if nombre.upper() in contactos:
        print("telefono", contactos[nombre.upper()])
    else:
        print("Contacto no encontrado. ")


def dividir_numeros():
    try:
        numero_uno = float(input("ingresa el primer número: "))
        numero_dos = float(input("Ingresa el segundo número: "))

        división = numero_uno / numero_dos

        print("Resultado: ", division)

    except ZeroDivisionError:
        print("ERROR: No se puede dividir entre cero")


    except ValueError:
        print("ERROR: Debes ingresar numeros ")

def analizar_mensaje():
    mensaje = input("ingresa un mensaje: ")

    print("longitud del mensaje:", len(mensaje))

    mayusculas = mensaje.upper()
    print("En mayusculas: ", mayusculas)

    palabras = mensaje.split()
    print("Total de palabras", len(palabras))

while True:

    print("\n-----------MENU PRINCIPAL----------")
    print("1. Mostrar numeros ordenados")
    print("2. Buscar telefono")
    print("3. Dividir dos numeros")
    print("4. Analizar mensaje") 
    print("5. Salir") 

    opcion = int(input("Selecciona una opcion (1-5):"))

    match opcion:

        case 1:
            ordenar_numeros()

        case 2:
            buscar_telefono()

        case 3:
            dividir_numeros()

        case 4:
            analizar_mensaje()

        case 5:
            print("Programa finalizado")
            break

        case _:
            print("Opcion no valida")
```
