ALGORITMO PARA CALCULAR LA VELOCIDAD 

```python
distancia = int(input("Ingrese la distancia en kilometros: "))
tiempo = int(input("Ingrese el tiempo en horas: "))

print("el dato distancia es de tipo: ", type(distancia))
print("el dato tiempo es de tipo: ", type(tiempo))
velocidad = distancia / tiempo
print("La velocidad promedio es:", velocidad, "km/h")

```
