Extra 1: Suma de todos los elementos de una matriz

En este ejercicio se creó una matriz de 3x3 y se utilizaron ciclos "for" para recorrer sus filas y elementos. La función "suma_matriz()" va sumando cada número y al final regresa el resultado con "return". También se creó una función para imprimir la matriz sin corchetes ni comas

```python
def suma_matriz(matriz):
    suma = 0

    for fila in matriz:
        for numero in fila:
            suma = suma + numero 

    return suma

def imprimir_matriz(matriz):
    for fila in matriz:
        for numero in fila:
            print(numero, end=" ")
        print()

matriz = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

imprimir_matriz(matriz)

resultado = suma_matriz(matriz)

print(f"EL RESULTADO DE TODOS LOS ELEMENTOS DE LA MATRIZ ES: {resultado}")
```

Extra 2: Matriz con suma por fila

En este ejercicio se recorrió una matriz para obtener la suma de cada una de sus filas, la función "suma_fila()" recibe una fila, suma sus elementos y regresa el resultado. Después se muestra la matriz y la suma correspondiente a cada fila.

```python
def suma_fila(fila):
    suma = 0

    for numero in fila:
        suma = suma + numero 
    return suma

def mostrar_tabla(matriz):
    for fila in matriz:
        for numero in fila:
            print(numero, end=" ")
        print()

matriz = [
    [3, 1, 4],
    [1, 5, 9],
    [2, 6, 5]
]

mostrar_tabla(matriz)

for i in range(len(matriz)):
    resultado = suma_fila(matriz[i])
    print("Suma de la fila", i, ":", resultado)
```

Extra 3: Producto de una columna sin usar el operador * (multiplicacion)

En este ejercicio se pidió al usuario seleccionar una columna de la matriz. Después se recorrieron los números de esa columna para obtener su producto. Como no se podía utilizar el operador de multiplicación, el cálculo se realizó mediante sumas repetidas usando ciclos "for"

```python
def multiplicar_columna(matriz, columna):
    producto = 1

    for fila in matriz:
        numero = fila[columna]

        resultado = 0

        for i in range(numero):
            resultado = resultado + producto

        producto = resultado

    return producto

matriz = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9],
    [2, 4, 6]
]

columna = int(input("INGRESA EL NUMERO DE COLUMNA 1, 2 o 3: "))
resultado = multiplicar_columna(matriz, columna)

print("EL PRODUCTO DE LA COLUMNA", columna,"ES", resultado)
```

Extra 4: Tabla de Pitágoras de tamaño variable

En este ejercicio se pidió al usuario un tamaño entre 2 y 5 para generar una tabla de Pitágoras. Los productos se calcularon mediante sumas repetidas, sin utilizar el operador de multiplicación. Al final, el usuario puede indicar un renglón y una columna para consultar el resultado almacenado en la tabla.

```python

def generar_tabla(tamaño):
    tabla = []

    for fila in range(1, tamaño + 1):
        nueva_fila = []

        for columna in range(1, tamaño + 1):
            resultado = 0
            for i in range(columna):
                resultado = resultado + fila


            nueva_fila.append(resultado)
    
        tabla.append(nueva_fila)

    return tabla

def imprimir_tabla(tabla):
    for fila in tabla:
        for numero in fila:
            print(numero,end="\t")
        print()

tamaño = int(input("INGRESE TAMAÑO DE LA TABLA (2-5):"))

if tamaño >= 2 and tamaño <= 5:
    tabla = generar_tabla(tamaño)
    imprimir_tabla(tabla)

    renglon = int(input("ingrese renglon:"))
    columna = int(input("ingresa columna:"))
    
    resultado = tabla[renglon - 1][columna - 1]

    print("el producto de", renglon, "x", columna, "es", resultado)

else:
    print("ERROR: EL TAMANO DEBE SER ENTRE 2 Y 5")
```
