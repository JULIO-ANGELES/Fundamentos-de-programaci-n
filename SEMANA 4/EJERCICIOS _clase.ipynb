```python
filas = 3
columnas = 3

matriz = []
for f in range (filas):
    fila_actual = []
    for c in range(columnas):
        valor = int(input(f"Valor para [{f}][{c}]"))
        fila_actual.append(valor)
    matriz.append(fila_actual)

print("impresion de matriz")
for f in range(filas):
    for c in range(columnas):
        print(matriz[f][c], end=" ")
    print()
```

```python
dimension = int(input("ingresa la dimension de la matriz: "))
i = 0

matriz = []
for f in range (dimension):
    fila_actual = []
    for c in range(columnas):
        valor = 0
        inv = dimension - i
        if c == inv:
            fila_actual.append(i)
        elif f == c:
             fila_actual.append(1)
        else:
            fila_actual.append(0)
            valor = 1
            i += 1

        fila_actual.append(valor)
        
    matriz.append(fila_actual)

print("impresion de matriz identidad:")
for f in range(dimension):
    for c in range(columnas):
        print(matriz[f][c], end=" ")
    print()
```

```python
lista = [] 
contador = 0
while contador < 10:
    valor = int(input("introduce un numero entre -10 y 10:"))
    if valor >= -10 and valor <= 10:
        contador += 1
        lista.append(valor)
    else:
        print("rango incorrecto, introduce un numero entre -10 y 10 ")
    
for i in lista:
    ordenar = 10
    asteriscos = i 
    asteriscos_negativos = asteriscos * -1
    ordenar_negativos = ordenar - asteriscos_negativos
    
    if i < 0:
        print(ordenar_negativos * " " + "*" * asteriscos_negativos + "|" + ordenar_negativos * " " )
        
    elif i > 0:
        print(ordenar * " " + "|" + asteriscos * "*")


```
