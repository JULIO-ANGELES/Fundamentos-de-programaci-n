Extra 1: División de cuenta con propina: Crea un programa que pida el total de la cuenta de un restaurante,
el porcentaje de propina a dejar y el número de personas que pagarán. El programa
debe calcular el monto de la propina, el total a pagar con propina y cuánto le toca
pagar a cada persona (con dos decimales).

```python
#JULIO CESAR ANGELES MENDOZA (IDS)
total_cuenta = float(input("introduzca el total de la cuenta:"))
personas = int(input("introduzca el numero de personas: "))
porcentaje_propina = float(input("introduzca el porcentaje de propina que gustarias dejar:"))

propina_final = total_cuenta * (porcentaje_propina / 100)
total_cuenta_final = total_cuenta + propina_final 
pago_correspondiente_por_persona = total_cuenta_final / personas 


print(f"el total de propina es: ${propina_final:.2f}")
print(f"el total de la cuenta con propina es: ${total_cuenta_final:.2f}")
print(f"el pago correspondiente por persona es: ${pago_correspondiente_por_persona:.2f}")
```
