Extra 1: División de cuenta con propina: Crea un programa que pida el total de la cuenta de un restaurante,
el porcentaje de propina a dejar y el número de personas que pagarán. El programa
debe calcular el monto de la propina, el total a pagar con propina y cuánto le toca
pagar a cada persona (con dos decimales).

```python
#JULIO CESAR ANGELES MENDOZA (IDS)
total_cuenta = float(input("introduzca el total de la cuenta:"))
personas = int(input("introduzca el numero de personas: "))
porcentaje_propina = float(input("introduzca el porcentaje de propina que gustarias dejar:"))

propina_final = total_cuenta * (porcentaje_propina / 100)
total_cuenta_final = total_cuenta + propina_final 
pago_correspondiente_por_persona = total_cuenta_final / personas 


print(f"el total de propina es: ${propina_final:.2f}")
print(f"el total de la cuenta con propina es: ${total_cuenta_final:.2f}")
print(f"el pago correspondiente por persona es: ${pago_correspondiente_por_persona:.2f}")
```

Extra 2: Conversor de minutos a días, horas y minutos
Crea un programa que pida una cantidad total de minutos (entero) y
la convierta a días, horas y minutos restantes. Utiliza los operadores de división
entera `//` y módulo `%`. (Pista: 1 día = 1440 minutos, 1 hora = 60 minutos.)

```python
#JULIO CESAR ANGELES MENDOZA (IDS)

minutos_totales = int(input("introduzca el total de minutos: "))
Dias = minutos_totales // 1440
restantes = minutos_totales % 1440
horas = restantes // 60
minutos = restantes % 60 

print(f"el total de dias es: {Dias} dias, {horas} horas y {minutos} minutos")
```

Extra 3: Calificación final ponderada
Crea un programa que pida las calificaciones de tres parciales
(valores de 0 a 10) y calcule la calificación final considerando una ponderación de
**30%, 30% y 40%** respectivamente. Muestra el resultado con dos decimales.

```python
#JULIO CESAR ANGELES MENDOZA (IDS)
primer_parcial = float(input("PARCIAL 1 (30%):"))
segundo_parcial = float(input("PARCIAL 2 (30%):"))
tercer_parcial = float(input("PARCIAL 3 (40%):"))

calificacion_final = (primer_parcial * 0.3) + (segundo_parcial * 0.3) + (tercer_parcial * 0.4)
print(f"tu calificacion final es: {calificacion_final:.2f}")
```

Extra 4: Conversor de moneda (MXN a USD y EUR)
Crea un programa que pida una cantidad en pesos mexicanos y los
tipos de cambio del dólar (USD) y del euro (EUR). Debe calcular y mostrar las
equivalencias redondeadas a dos decimales. (Fórmula: `cantidad / tipo_de_cambio`.)

```python
#JULIO CESAR ANGELES MENDOZA (IDS)
mxn = float(input("introduzca la cantidad en pesos mexicanos (MXN): "))
tipo_cambio_usd = float(input("Tipo de cambio USD: "))
tipo_cambio_eur = float(input("Tipo de cambio EUR: "))

usd = mxn / tipo_cambio_usd
eur = mxn / tipo_cambio_eur

print(f"${mxn:.2f} MXN equivalen a:")
print(f"USD: - ${usd:.2f}")
print(f"EUR: - €{eur:.2f}")    
```
