```python
import requests

url = "https://www.google.com"

try:
    respuesta = requests.get(url)

    if respuesta.status_code == 200:
        print("La página web está disponible.")
    else:
        print(f"Estado: {respuesta.status_code}")

except Exception as e:
    print(f"Error: {e}")
```

```python
!pip install psutil
import psutil
import matloplib.pyplot as plt

cpu = psutil.cpu_percent()
ram = psutil.virtual_memory().percent

metricas = ['CPU', 'RAM']
valores = [cpu, ram]

plt.bar(metricas, valores)
plt.title('Uso de CPU y RAM')
plt.ylabel('monitoreos de recursos')
plt.ylim(0,100)

plt.show()
```
